```python
import pandas as pd
import os

# Tentukan path folder raw yang berisi file CSV
base_path = '/content/drive/MyDrive/0_projek/ecommerce-delivery-analysis/data/raw/'

# Daftar nama file yang akan dianalisis
file_names = [
    'olist_orders_dataset.csv',
    'olist_order_items_dataset.csv',
    'olist_customers_dataset.csv',
    'olist_sellers_dataset.csv',
    'olist_products_dataset.csv',
    'olist_order_reviews_dataset.csv',
    'olist_order_payments_dataset.csv'
]

# Dictionary untuk menyimpan dataframe
datasets = {}

# Fungsi untuk membaca dan menampilkan summary data
for file in file_names:
    file_path = os.path.join(base_path, file)

    # Membaca dataset
    df = pd.read_csv(file_path)

    # Menyimpan dataframe ke dalam dictionary dengan nama file sebagai key (tanpa .csv)
    name = file.replace('.csv', '')
    datasets[name] = df

    # Menampilkan informasi dataset
    print(f"========== TABEL {name.upper()} ==========")
    print(f"Jumlah Baris adalah {df.shape[0]}")
    print(f"Jumlah Kolom adalah {df.shape[1]}")
    print("Daftar Kolom dan Tipe Data:")
    print(df.dtypes)
    print("\nContoh 3 Baris Pertama:")
    display(df.head(3))
    print("\n" + "="*50 + "\n")
```

```python
import pandas as pd

# 1. Membuat daftar untuk menyimpan hasil asesmen
dq_summary = []

# 2. Melakukan iterasi pada setiap tabel di dalam dictionary
for name, df in datasets.items():
    # Menghitung total missing values di seluruh kolom tabel
    total_missing = df.isnull().sum().sum()

    # Menghitung total baris yang terduplikasi secara identik
    total_duplicates = df.duplicated().sum()

    # Menyimpan hasil ke dalam daftar
    dq_summary.append([name, df.shape[0], total_missing, total_duplicates])

# 3. Mengubah hasil menjadi DataFrame agar rapi
df_dq = pd.DataFrame(dq_summary, columns=['Nama Tabel', 'Total Baris', 'Total Missing Values', 'Total Duplikat'])

print("Laporan Kualitas Data Utama")
display(df_dq)

print("\n")

# 4. Mengecek spesifik missing values pada tabel orders
print("Rincian Missing Values pada olist_orders_dataset")
display(datasets['olist_orders_dataset'].isnull().sum())

print("\n")

# 5. Mengecek tipe data tanggal di tabel orders
print("Tipe data kolom tanggal di olist_orders_dataset")
print(datasets['olist_orders_dataset'][['order_purchase_timestamp', 'order_delivered_customer_date']].dtypes)
```

```python
import pandas as pd
import os

# 1. Mengambil salinan dataframe dari dictionary sebelumnya untuk menjaga data asli
df_orders = datasets['olist_orders_dataset'].copy()
df_items = datasets['olist_order_items_dataset'].copy()
df_reviews = datasets['olist_order_reviews_dataset'].copy()
df_customers = datasets['olist_customers_dataset'].copy()
df_sellers = datasets['olist_sellers_dataset'].copy()

# 2. Menyaring hanya pesanan dengan status delivered
df_orders_delivered = df_orders[df_orders['order_status'] == 'delivered'].copy()

# 3. Menghapus missing values pada kolom waktu tiba
# Terdapat segelintir anomali berstatus delivered namun waktu tibanya kosong
df_orders_delivered = df_orders_delivered.dropna(subset=['order_delivered_customer_date'])

# 4. Mengubah kolom waktu menjadi datetime
waktu_kolom = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in waktu_kolom:
    df_orders_delivered[col] = pd.to_datetime(df_orders_delivered[col])

# 5. Membuat feature baru untuk evaluasi pengiriman
# Menghitung durasi pengiriman aktual dalam hari
df_orders_delivered['actual_delivery_days'] = (df_orders_delivered['order_delivered_customer_date'] - df_orders_delivered['order_purchase_timestamp']).dt.days

# Menghitung durasi pengiriman estimasi dalam hari
df_orders_delivered['estimated_delivery_days'] = (df_orders_delivered['order_estimated_delivery_date'] - df_orders_delivered['order_purchase_timestamp']).dt.days

# Menghitung selisih keterlambatan dalam hari
df_orders_delivered['delay_days'] = (df_orders_delivered['order_delivered_customer_date'] - df_orders_delivered['order_estimated_delivery_date']).dt.days

# Membuat indikator biner untuk pesanan terlambat (1 bernilai terlambat, 0 bernilai tepat waktu)
df_orders_delivered['is_late'] = df_orders_delivered['delay_days'].apply(lambda x: 1 if x > 0 else 0)

# 6. Menggabungkan tabel menjadi tabel master
# Menggabungkan pesanan dengan pelanggan
df_master = pd.merge(df_orders_delivered, df_customers, on='customer_id', how='inner')

# Menggabungkan dengan item pesanan
df_master = pd.merge(df_master, df_items, on='order_id', how='inner')

# Menggabungkan dengan penjual
df_master = pd.merge(df_master, df_sellers, on='seller_id', how='inner')

# Menggabungkan dengan ulasan pelanggan
# Satu pesanan dapat terdiri dari beberapa item fisik sehingga tabel ulasan harus dihitung rata-ratanya terlebih dahulu agar tidak terjadi duplikasi baris saat digabungkan
review_summary = df_reviews.groupby('order_id')['review_score'].mean().reset_index()
df_master = pd.merge(df_master, review_summary, on='order_id', how='left')

# 7. Menyimpan dataset yang telah bersih
# Membuat folder processed di Google Drive jika belum tersedia
processed_path = '/content/drive/MyDrive/0_projek/ecommerce-delivery-analysis/data/processed/'
if not os.path.exists(processed_path):
    os.makedirs(processed_path)

file_simpan = os.path.join(processed_path, 'master_dataset.csv')
df_master.to_csv(file_simpan, index=False)

print(f"Data bersih berhasil disimpan di {file_simpan}")
print(f"Jumlah baris tabel master adalah {df_master.shape[0]}")
print(f"Jumlah kolom tabel master adalah {df_master.shape[1]}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Memuat data bersih
file_path = '/content/drive/MyDrive/ecommerce-delivery-analysis/data/processed/master_dataset.csv'
df = pd.read_csv(file_path)

# Tabel master berbasis item, oleh sebab itu kamu perlu menghapus duplikasi order_id untuk analisis level pesanan
df_orders = df.drop_duplicates(subset=['order_id']).copy()

# A. Delivery Performance
late_counts = df_orders['is_late'].value_counts(normalize=True) * 100
rata_waktu_kirim = df_orders['actual_delivery_days'].mean()
rata_keterlambatan = df_orders[df_orders['is_late'] == 1]['delay_days'].mean()

print("=== RINGKASAN PERFORMA PENGIRIMAN ===")
print(f"Pesanan Tepat Waktu adalah {late_counts[0]:.2f}%")
print(f"Pesanan Terlambat adalah {late_counts[1]:.2f}%")
print(f"Rata-rata Waktu Pengiriman adalah {rata_waktu_kirim:.2f} hari")
print(f"Rata-rata Lama Keterlambatan adalah {rata_keterlambatan:.2f} hari\n")

# B. Customer Satisfaction
plt.figure(figsize=(8, 5))
sns.barplot(x='is_late', y='review_score', data=df_orders, errorbar=None, palette='viridis')
plt.title('Rata-Rata Rating Pelanggan berdasarkan Status Keterlambatan')
plt.xlabel('Status (0 = Tepat Waktu, 1 = Terlambat)')
plt.ylabel('Rata-Rata Skor Ulasan')
plt.ylim(0, 5)
plt.show()
```

```python
# C. Geographical Analysis
state_delay = df_orders.groupby('customer_state')['is_late'].mean().sort_values(ascending=False) * 100

plt.figure(figsize=(14, 6))
sns.barplot(x=state_delay.index, y=state_delay.values, palette='Reds_r')
plt.title('Persentase Keterlambatan Pengiriman per Negara Bagian (Customer)')
plt.xlabel('Negara Bagian (State)')
plt.ylabel('Persentase Keterlambatan (%)')
plt.show()

# D & E. Transaction & Shipping Analysis (Hubungan Biaya dan Jarak)
plt.figure(figsize=(14, 6))
sns.boxplot(x='customer_state', y='freight_value', data=df_orders, order=state_delay.index, showfliers=False)
plt.title('Distribusi Biaya Pengiriman (Freight Value) per Negara Bagian')
plt.xlabel('Negara Bagian (Diurutkan dari keterlambatan tertinggi)')
plt.ylabel('Biaya Pengiriman')
plt.show()
```

```python
# F. Time-Based Analysis
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'])
df_orders['year_month'] = df_orders['order_purchase_timestamp'].dt.to_period('M')

trend_delay = df_orders.groupby('year_month')['is_late'].mean() * 100

plt.figure(figsize=(14, 6))
trend_delay.plot(kind='line', marker='o', color='crimson')
plt.title('Tren Persentase Keterlambatan per Bulan (2016 - 2018)')
plt.xlabel('Bulan Pembelian')
plt.ylabel('Tingkat Keterlambatan (%)')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import pandas_gbq
from google.cloud import bigquery
from google.cloud.exceptions import NotFound

# 1. Masukkan Project ID dari Google Cloud Console
project_id = 'fourth-cirrus-514014-a8'
nama_dataset = 'ecommerce_olist'
dataset_id = f"{project_id}.{nama_dataset}"

# 2. Inisiasi client BigQuery dan pembuatan dataset otomatis
client = bigquery.Client(project=project_id)

try:
    client.get_dataset(dataset_id)
    print(f"Dataset {nama_dataset} sudah tersedia.")
except NotFound:
    print(f"Dataset {nama_dataset} belum ada. Sistem sedang membuat dataset baru...")
    dataset = bigquery.Dataset(dataset_id)
    dataset.location = "US"
    client.create_dataset(dataset, timeout=30)
    print("Dataset berhasil dibuat.")

# 3. Mengunggah data menggunakan pandas_gbq
print("Sedang mengunggah data ke BigQuery...")
try:
    pandas_gbq.to_gbq(
        df_master,
        destination_table=f'{nama_dataset}.master_orders',
        project_id=project_id,
        if_exists='replace',
        progress_bar=True
    )
    print("Berhasil! Data telah tersimpan di tabel BigQuery.")
except Exception as e:
    print("Terjadi kesalahan spesifik saat mengunggah data:")
    print(e)
```
